```python
from langgraph.graph import StateGraph,START,END
import os
from dotenv import load_dotenv
from typing import TypedDict
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
# Initialize Gemini
load_dotenv()
llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    google_api_key=os.getenv("GOOGLE_API_KEY")
)
```

```python
class LLMState(TypedDict): 
    qustions:str
    answer:str
    
```

```python
def llmnode(state:LLMState)-> LLMState:
    question = state['qustions']
    prompt = f"Answer the following question {question}"
    response=llm.invoke(prompt).content
    state['answer'] = response
    return state
```

```python
graph  = StateGraph(LLMState)

graph.add_node('llmmode',llmnode)

graph.add_edge(START,'llmmode')
graph.add_edge('llmmode',END)

workflow = graph.compile()

```

```python
final_state=workflow.invoke({'qustions':"what is use of LangGraph"})
print(final_state)
```
